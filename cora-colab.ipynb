## Célula 2 — Configurar a API

```python
import google.generativeai as genai
from google.colab import userdata

# Lê a chave do painel de Secrets do Colab
api_key = userdata.get('GEMINI_API_KEY')
genai.configure(api_key=api_key)

# Lê as instruções da Cora
with open('chatbot-atendimento-casa-cor/02-instrucoes-v5.md', 'r', encoding='utf-8') as f:
    system_prompt = f.read()

# Cria o modelo com as instruções
model = genai.GenerativeModel(
    model_name='gemini-3.8-flash',
    system_instruction=system_prompt
)

print('✅ Cora pronta para atender!')
```

## Célula 3 — Iniciar o chat

Execute a célula abaixo. Digite suas mensagens no campo que aparecer e pressione **Enter**. Para encerrar, digite `sair`.

```python
chat = model.start_chat(history=[])

print('=' * 50)
print('  Cora — Assistente Virtual da Casa & Cor')
print('  (digite "sair" para encerrar)')
print('=' * 50)

while True:
    entrada = input('\nVocê: ').strip()

    if entrada.lower() == 'sair':
        print('\nCora: Até mais! 👋')
        break

    if not entrada:
        continue

    resposta = chat.send_message(entrada)
    print(f'\nCora: {resposta.text}')
```
